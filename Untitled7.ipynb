```python
students = {
    "Ali": 85,
    "Sara": 72,
    "Ahmed": 35,
    "Ayesha": 91,
    "Usman": 55
}

def calculate_grade(marks):
    if marks >= 80:
        return "A"
    elif marks >= 60:
        return "B"
    elif marks >= 40:
        return "C"
    else:
        return "Fail"

def count_passed_students():
    count = 0
    for marks in students.values():
        if marks >= 40:
            count += 1
    return count

for name, marks in students.items():
    grade = calculate_grade(marks)
    print(f"Name: {name}, Marks: {marks}, Grade: {grade}")

print(f"Total Passed Students: {count_passed_students()}")
```

task *2*

```python
products = {
    "Milk": 250,
    "Bread": 150,
    "Eggs": 400,
    "Rice": 1200
}

def calculate_discount(total):
    if total >= 5000:
        return total * 0.10
    else:
        return 0

total = 0

for item, price in products.items():
    print(item, price)

while True:
    item = input("Enter product name or 'done' to finish: ").title()

    if item == "Done":
        break
    elif item in products:
        total += products[item]
        print(f"{item} added. Price: Rs. {products[item]}")
    else:
        print("Product not found.")

discount = calculate_discount(total)
final_bill = total - discount

print(f"Original Bill: Rs. {total}")
print(f"Discount: Rs. {discount}")
print(f"Final Bill: Rs. {final_bill}")
```

```python
account = {
    "name": "Ali",
    "pin": 1234,
    "balance": 10000
}

def withdraw(balance, amount):
    if amount <= balance:
        balance -= amount
        print("Withdrawal successful.")
        print(f"Remaining balance: Rs. {balance}")
        return balance
    else:
        print("Insufficient balance.")
        return balance

attempts = 0

while attempts < 3:
    pin = int(input("Enter your PIN: "))

    if pin == account["pin"]:
        print(f"Welcome, {account['name']}!")

        amount = int(input("Enter withdrawal amount: "))

        account["balance"] = withdraw(account["balance"], amount)

        break
    else:
        attempts += 1
        print("Incorrect PIN.")

        if attempts == 3:
            print("Your account is locked due to 3 incorrect attempts.")
```

# ***task4***

```python
def check_guess(guess, secret):
    if guess == secret:
        return "Correct!"
    elif guess < secret:
        return "Too low!"
    else:
        return "Too high!"


secret_number = 7
attempts = 0

player = {
    "name": input("Enter your name: "),
    "score": 0
}

while attempts < 5:
    guess = int(input("Guess the number: "))
    attempts += 1

    result = check_guess(guess, secret_number)
    print(result)

    if result == "Correct!":
        player["score"] = 100 - (attempts - 1) * 20
        print(f"Congratulations, {player['name']}!")
        break

if attempts == 5 and guess != secret_number:
    print("Game over! You ran out of attempts.")

print(f"Player: {player['name']}")
print(f"Score: {player['score']}")
```

# ***task5***

```python
menu = {
    "Biryani": 350,
    "Karahi": 1200,
    "Naan": 50,
    "Tea": 100
}

def calculate_item_total(price, quantity):
    return price * quantity

total = 0
orders = {}

for item, price in menu.items():
    print(item, "Rs.", price)

while True:
    item = input("Enter item name or done: ").title()

    if item == "Done":
        break

    if item in menu:
        quantity = int(input("Enter quantity: "))

        if quantity > 0:
            item_total = calculate_item_total(menu[item], quantity)
            total += item_total

            if item in orders:
                orders[item] += quantity
            else:
                orders[item] = quantity

            print(item, "Total: Rs.", item_total)
        else:
            print("Invalid quantity.")
    else:
        print("Item not found in menu.")

print("\nOrdered Items:")

for item, quantity in orders.items():
    print(item, "Quantity:", quantity)

print("Total bill: Rs.", total)
```

## *task6*

```python
employees = {
    "Ali": 26,
    "Sara": 23,
    "Ahmed": 18,
    "Ayesha": 30
}

def calculate_salary(days):
    if days == 26:
        return 30000
    elif days >= 20 and days <= 25:
        return 30000 * 0.80
    elif days < 20:
        return 30000 * 0.50
    else:
        return "Invalid attendance"

full_salary_count = 0

for name, days in employees.items():
    salary = calculate_salary(days)

    if salary == 30000:
        full_salary_count += 1

    print("Employee:", name)
    print("Attendance:", days)

    if salary == "Invalid attendance":
        print(salary)
    else:
        print("Salary: Rs.", salary)

    print("------------------")

print("Employees receiving full salary:", full_salary_count)
```
